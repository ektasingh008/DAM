```python
#Byjus case
import pandas as pd
import numpy as np

df = pd.read_excel("/content/Lecture2_data (2).xlsx")
```

```python
#Data Description
print(df.head)
print(df.shape)
df.info()
```

```python
df.dtypes
```

```python
#Top 10 profit/loss product
df["Total_Cost_Lakh"] = (
    df["Marketing_Spend_Lakh"]
    + df["Content_Cost_Lakh"]
    +df["Salary_Cost_Lakh"]
    +df["Other_Cost_Lakh"]
)
df[ "Profit_Lakh"] = df ["Revenue_Lakh"] - df["Total_Cost_Lakh"]
print(df[["Product", "Revenue_Lakh","Total_Cost_Lakh","Profit_Lakh"]].head(10))
```

```python
#Missing Value
print(df.isna().sum())
```

```python
# Decision making structures
#If Statements
rating = df["Revenue_Lakh"].mean()

if rating >= 4:
  print("revenue is good")
```

```python
# Decision making structures
#If else
revenue = df["Revenue_Lakh"].mean()

if revenue >= 300:
  print("Average revenue is high")
else:
    print("Average revenue is low")
```

```python
# Decision making structures
# Nested IF Statement in python
revenue = df["Revenue_Lakh"].mean()
rating = df["Customer_Rating"].mean()

if revenue >= 300:
  print("Revenue is high")

  if rating >= 4:
     print("Customer satisfaction is also good")
  else:
      print ("Customer satisfaction needs improvement")

else:
    print("Revenue is low")
```

```python
#WHILE Loop in python
#While loop and i,j is variable to track
i = 0

while i < 5:
  revenue = df["Revenue_Lakh"].iloc[i]

  if revenue >= 300:
      print("High Revenue:", revenue)
  else :
      print("Low Revenue:", revenue)
```

```python
#Starbucks Case
import pandas as pd
import numpy as np

# Define the correct file path
file_path = "/content/Starbucks_India_Sales_Dataset.xlsx"

# Load the correct Excel sheets
sales_df = pd.read_excel(file_path, sheet_name="Sales_Data")
products_table = pd.read_excel(file_path, sheet_name="Products")
stores_df = pd.read_excel(file_path, sheet_name="Stores")

# Check the data
print("Sales table:", sales_df.shape)
print("Products:", products_table.shape)
print("Stores:", stores_df.shape)

# Display first 5 rows of stores
stores_df.head()
```

```python
# ---- Part A: create them by typing ----

# LIST: the Starbucks menu (8 products)
menu = ["Cappuccino", "Latte", "Americano", "Cold Brew",
        "Frappuccino", "Masala Chai", "Blueberry Muffin", "Veg Sandwich"]

# TUPLE: one store's fixed details (StoreID, name, city)
store = ("S01", "Connaught Place", "Delhi")

# SET: cities where Starbucks sells (the duplicate 'Delhi' is removed automatically)
cities = {"Delhi", "Mumbai", "Bengaluru", "Hyderabad", "Pune", "Delhi"}

# DICTIONARY: product -> base price in Rs
price = {"Cappuccino": 250, "Latte": 270, "Americano": 220, "Cold Brew": 300,
         "Frappuccino": 350, "Masala Chai": 200, "Blueberry Muffin": 180, "Veg Sandwich": 240}

print("List      :", menu)
print("Tuple     :", store)
print("Set       :", cities, " <- 'Delhi' typed twice, stored once")
print("Dictionary:", price)
print()
print(type(menu), type(store), type(cities), type(price))
```

```python
# ---- Part B: create them from the Excel data ----
# Each Excel column becomes a Python list. Each Excel row becomes a dictionary.

product_list = sales_df["Product"].tolist()        # 100 products (with repeats)
city_list    = sales_df["City"].tolist()           # 100 cities  (with repeats)
qty_list     = sales_df["Quantity"].tolist()       # 100 quantities
revenue_list = sales_df["Revenue"].tolist()        # 100 revenue values

first_order   = sales_df.iloc[0].to_dict()                 # ONE order  -> dictionary
first_5_orders = sales_df.head(5).to_dict("records")       # FIVE orders -> list of dictionaries

print("Items in product_list :", len(product_list))
print("Unique cities (set)   :", set(city_list))
print()
print("First order as a dictionary:")
print(first_order)
```

```python
# --- List ---
print("First item      :", menu[0])
print("Last item       :", menu[-1])
print("First three     :", menu[0:3])      # slicing: start included, end excluded
print("Every 2nd item  :", menu[::2])

# --- Tuple (same as list) ---
print("\nStore ID        :", store[0])
print("Store city      :", store[2])

# --- Dictionary ---
print("\nPrice of Latte  :", price["Latte"])
print("Safe lookup     :", price.get("Pizza", "Not on menu"))   # no error if key is missing

# --- Set ---
print("\nIs Pune a Starbucks city?   ", "Pune" in cities)
print("Is Jaipur a Starbucks city? ", "Jaipur" in cities)
```
